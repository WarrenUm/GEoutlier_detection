## Exercise 1
A fast-food chain claims that the mean time to order food at their restaurants is 60 seconds, with a standard deviation of 30 seconds. You decide to put this claim to the test and go to one of the restaurants to observe actual waiting times. You take a sample of 36 customers and find that the mean order time was 75 seconds. Does this finding provide enough evidence to contradict the fast food chain's claim of fast service?

Follow the 5 steps shown in previous lesson and use $\alpha$ = 0.05. 

```python
# State your null and alternative hypotheses
#null: mean time to order food it greater than 60 sec
#alt: mean time is less than or equal to 60 sec
```

```python
# Your solution here
mu = 60 #popmean
sig = 30 #sd of pop
n = 36 #sample size
x_bar = 75 #sample mean
alpha = 0.05

import scipy.stats as stats
from math import sqrt

z = (x_bar - mu) / (sig/sqrt(n))
z
p = 1 - stats.norm.cdf(z)
p
# (p = 0.0013498980316301035, z = 3.0)
```

```python
#plot
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('seaborn')
plt.fill_between(x=np.arange(-3,3.0,0.05),
                y1 = stats.norm.pdf(np.arange(3,3,0.05)),
                facecolor='red',
                alpha = 0.35,
                label='Area below z-stat')

plt.fill_between(x=np.arange(3.0,3,0.05),
                y1 = stats.norm.pdf(np.arange(-3,3,0.05)),
                facecolor='blue',
                alpha = 0.35,
                label='Area above z-stat')
```

## Exercise 2

25 students complete a preparation program for taking the SAT test.  Here are the SAT scores from the 25 students who completed the program:

``
434 694 457 534 720 400 484 478 610 641 425 636 454 
514 563 370 499 640 501 625 612 471 598 509 531
``

We know that the population average for SAT scores is 500 with a standard deviation of 100.

Are our 25 students’ SAT scores significantly higher than the population's mean score? 

*Note that the SAT preparation program claims that it will increase (and not decrease) the SAT score.  So, you can conduct a one-directional test. (alpha = .05).*

```python
# State your hypotheses 
#null: our students scores are not signuficantly higher than the mean
#alt: out students scores are sig higher than the mean
```

```python
# Give your solution here 
mu = 500 #popmean
sig = 100 #sd of pop
n = 25 #sample size
scores = [434, 694, 457, 534, 720, 400, 484, 478, 610, 641, 425, 636, 454, 514, 563, 370, 499, 640, 501, 625, 612, 471, 598, 509, 531]

x_bar = np.mean(scores) #sample mean
alpha = 0.05

import scipy.stats as stats
from math import sqrt

z = (x_bar - mu) / (sig/sqrt(n))
z
p = 1 - stats.norm.cdf(z)
p

p,z
# p = 0.03593031911292577, z = 1.8
```

```python
# Interpret the results in terms of the p-value
#the p value is less than 0.05 so we reject the null in favor of the alt
```
